# Ασκήσεις εξάσκησης

Σε αυτά τα τετράδια θα βρεις μικρά προγράμματα όπου λείπει ένα κομμάτι κώδικα.
Το κομμάτι που λείπει σημειώνεται με `___` (τρεις παύλες).

- Διάβασε την εκφώνηση πάνω από κάθε κελί και αντικατάστησε το `___` με το
  σωστό κομμάτι κώδικα — μπορεί να είναι μια εντολή (`input`, `int`, `float`),
  μια μεταβλητή, ένα μήνυμα σε εισαγωγικά ή ολόκληρη έκφραση.
- Τρέξε το κελί με **Shift+Enter** και έλεγξε ότι η έξοδος ταιριάζει με το
  αναμενόμενο που γράφει η εκφώνηση.
- Αν τρέξεις το κελί χωρίς να συμπληρώσεις το `___`, η Python θα παραπονεθεί
  ότι δεν ξέρει τι είναι — φυσιολογικό, απλώς δεν το έχεις συμπληρώσει ακόμα.

Στο σετ αυτό, κάθε κελί κώδικα έχει **ένα** κενό.
