# Εξάσκηση: `int()` ή `float()`;

- `int(...)` → για **ακέραιους** αριθμούς (χωρίς δεκαδικά): ηλικία, πλήθος, έτος...
- `float(...)` → για **δεκαδικούς**: τιμή, ύψος, θερμοκρασία, βαθμός...
- Τίποτα → όταν θέλουμε **κείμενο**: όνομα, πόλη, τηλέφωνο (ακόμα κι αν έχει ψηφία, δεν κάνουμε πράξεις μαζί του).

### Άσκηση 1 — Ηλικία

Διάβασε την ηλικία (ακέραιος) και εμφάνισε πόσα χρόνια θα έχει του χρόνου.

Συμπλήρωσε το `___`.

*Αναμενόμενο:* για είσοδο `15` τυπώνει `Του χρόνου θα είσαι 16`

```python
ilikia = ___(input("Ηλικία: "))
print("Του χρόνου θα είσαι", ilikia + 1)
```

### Άσκηση 2 — Ύψος

Διάβασε το ύψος σε μέτρα (δεκαδικός) και εμφάνισέ το σε εκατοστά.

Συμπλήρωσε το `___`.

*Αναμενόμενο:* για είσοδο `1.72` τυπώνει `Δηλαδή 172.0 cm`

```python
ypsos = ___(input("Ύψος σε m: "))
print("Δηλαδή", ypsos * 100, "cm")
```

### Άσκηση 3 — Τηλέφωνο

Διάβασε έναν αριθμό τηλεφώνου. Έχει ψηφία, αλλά **δεν κάνουμε πράξεις** μαζί του — πρέπει να μείνει κείμενο. Συμπλήρωσε τη σωστή μετατροπή.

Συμπλήρωσε το `___`.

*Αναμενόμενο:* για είσοδο `2410555010` τυπώνει `Το τηλέφωνό σου είναι 2410555010`

```python
tilefono = ___(input("Τηλέφωνο: "))
print("Το τηλέφωνό σου είναι", tilefono)
```

### Άσκηση 4 — Τιμή προϊόντος

Διάβασε την τιμή ενός προϊόντος (δεκαδικός) και εμφάνισε τη διπλή.

Συμπλήρωσε το `___`.

*Αναμενόμενο:* για είσοδο `12.5` τυπώνει `Διπλή τιμή: 25.0`

```python
timi = ___(input("Τιμή: "))
print("Διπλή τιμή:", timi * 2)
```

### Άσκηση 5 — Μέσος όρος δύο βαθμών

Ο πρώτος βαθμός διαβάζεται ήδη σωστά. Συμπλήρωσε το δεύτερο με τον ίδιο τρόπο (δεκαδικός).

Συμπλήρωσε το `___`.

*Αναμενόμενο:* για είσοδοι `17` και `18` τυπώνει `Μέσος όρος: 17.5`

```python
v1 = float(input("Βαθμός 1: "))
v2 = ___(input("Βαθμός 2: "))
mo = (v1 + v2) / 2
print("Μέσος όρος:", mo)
```

### Άσκηση 6 — Ακέραιοι σε πράξη

Διάβασε πόσα μολύβια αγόρασε ο μαθητής (ακέραιος) και υπόλογισε το κόστος (0.5€ το ένα).

Συμπλήρωσε το `___`.

*Αναμενόμενο:* για είσοδο `4` τυπώνει `Κόστος: 2.0 €`

```python
molivia = ___(input("Πόσα μολύβια: "))
print("Κόστος:", molivia * 0.5, "€")
```
