# Εκφώνηση

Το χιλιόμετρο (σύμβολο: km ή χλμ.) είναι μονάδα μέτρησης του μήκους και
χρησιμοποιείται για την μέτρηση μεγάλων αποστάσεων σε όλες σχεδόν τις χώρες
διεθνώς, με μοναδικές εξαιρέσεις τις ΗΠΑ και το Ηνωμένο Βασίλειο, όπου
χρησιμοποιείται το μίλι ξηράς.

Για να μετατρέψουμε τα χιλιόμετρα σε μίλια χρησιμοποιούμε τον εξής τύπο:
$ mi = 0.62 \cdot km $,
όπου $km$ είναι ο αριθμός των χιλιομέτρων.
